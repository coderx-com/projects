### The only difference is the cluster configuration, which uses one master node and two worker nodes. To control variables when comparing scaling-out training time, both clusters use the same base configuration of 2 vCPUs (1 core) and 8 GB memory per node.

```python
from pyspark.ml.regression import LinearRegression
from pyspark.ml.evaluation import RegressionEvaluator
import time

# Evaluators
rmse_evaluator = RegressionEvaluator(
    labelCol="price",
    predictionCol="prediction",
    metricName="rmse"
)

r2_evaluator = RegressionEvaluator(
    labelCol="price",
    predictionCol="prediction",
    metricName="r2"
)

ml_ready_df = spark.read.parquet("gs://jyairbnb/lr_ready/")
```

```python
# 10%
sample_10 = ml_ready_df.sample(False, 0.10, seed=123)
train_10, test_10 = sample_10.randomSplit([0.8, 0.2], seed=123)

lr = LinearRegression(
    featuresCol="features",
    labelCol="price",
    maxIter=50,
    regParam=0.1,
    elasticNetParam=0.5
)

start = time.time()
lr_model = lr.fit(train_10)
t10 = time.time() - start

pred10 = lr_model.transform(test_10)
rmse10 = rmse_evaluator.evaluate(pred10)
r2_10 = r2_evaluator.evaluate(pred10)

print("10% Runtime:", t10)
print("10% RMSE:", rmse10)
print("10% R^2:", r2_10)
```
