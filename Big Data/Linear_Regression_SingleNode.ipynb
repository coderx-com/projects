# This notebook shows the training results of Linear Regression when I run the codes in a sigle node cluster.

```python
from pyspark.ml.regression import LinearRegression
from pyspark.ml.evaluation import RegressionEvaluator
import time   

evaluator = RegressionEvaluator(
    labelCol="price",             # Target Variable
    predictionCol="prediction",
    metricName="rmse"             # Use RMSE as evaluation metric
)

ml_ready_df = spark.read.parquet("gs://jyairbnb/lr_ready/")
```

## Scaling Up: 10% - 25% - 50% - 100%

The model parameters were selected to support fast training and stable convergence. 

Since a simple linear model is unlikely to fully capture complex price patterns, linear regression is used as a baseline model. Its results provide a reference point for performance and runtime comparison with more advanced models such as Gradient Boosting Trees.

```python
# 10%
sample_10 = ml_ready_df.sample(False, 0.10, seed=123)

# Split sampled data into training and testing sets
train_10, test_10 = sample_10.randomSplit([0.8, 0.2], seed=123)

# Initialize the model
lr = LinearRegression(
    featuresCol="features",
    labelCol="price",
    maxIter=50,             
    regParam=0.1,
    elasticNetParam=0.5
)

# Train model and record runtime
start = time.time()
lr_model = lr.fit(train_10)
t10 = time.time() - start

# Generate predictions on test set
pred10 = lr_model.transform(test_10)

# Evaluate model using RMSE
rmse10 = evaluator.evaluate(pred10)

print("10% Runtime:", t10)
print("10% RMSE:", rmse10)
```

```python
# 25%
sample_25 = ml_ready_df.sample(False, 0.25, seed=123)
train_25, test_25 = sample_25.randomSplit([0.8, 0.2], seed=123)

lr = LinearRegression(
    featuresCol="features",
    labelCol="price",
    maxIter=50,
    regParam=0.1,
    elasticNetParam=0.5
)

start = time.time()
lr_model = lr.fit(train_25)
t25 = time.time() - start

pred25 = lr_model.transform(test_25)
rmse25 = evaluator.evaluate(pred25)

print("25% Runtime:", t25)
print("25% RMSE:", rmse25)
```
